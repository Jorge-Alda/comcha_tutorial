```python
import alpaca
import numpy as np
```

```python
def get_cbs(x):
  fa = 10**x[0]
  cqL33, cuR, cdR, clL33, ceR = x[1:] * fa
  couplings = alpaca.ALPcouplings({'cqL': np.diag([0,0,cqL33]), 'cuR': cuR, 'cdR': cdR, 'clL': np.diag([0,0,clL33]), 'ceR': ceR}, basis='derivative_above', scale=4*np.pi*fa).match_run(2.0, 'VA_below', integrator='leadinglog')
  return np.real(couplings['cdV'][1,2]-couplings['cdA'][1,2])

def get_cmumu(x):
  fa = 10**x[0]
  cqL33, cuR, cdR, clL33, ceR = x[1:] * fa
  couplings = alpaca.ALPcouplings({'cqL': np.diag([0,0,cqL33]), 'cuR': cuR, 'cdR': cdR, 'clL': np.diag([0,0,clL33]), 'ceR': ceR}, basis='derivative_above', scale=4*np.pi*fa).match_run(2.0, 'VA_below', integrator='leadinglog')
  return np.real(couplings['ceA'][1,1])
```
